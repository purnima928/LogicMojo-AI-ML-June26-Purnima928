```python
missing = pd.DataFrame({
    "Missing Count": app_df.isnull().sum(),
    "Missing %": round(app_df.isnull().mean()*100, 2)
})

missing = missing[missing["Missing Count"] > 0]
missing = missing.sort_values("Missing %", ascending=False)

missing
```

```python
plt.figure(figsize=(10,8))

missing["Missing %"].plot(kind="barh")

plt.xlabel("Missing Percentage")
plt.ylabel("Features")
plt.title("Missing Values (%)")
plt.yticks(fontsize=8)
plt.tight_layout()

plt.show()
```

```python
missing["Category"] = pd.cut(
    missing["Missing %"],
    bins=[0,10,30,50,100],
    labels=["Low (<10%)",
            "Moderate (10-30%)",
            "High (30-50%)",
            "Very High (>50%)"]
)

missing.head(30)
```

```python
missing["Category"].value_counts()
```

A large number of variables describing building and housing characteristics (e.g., common area, living apartments, elevators, land area) have more than 50% missing values. Since these variables are highly incomplete and many represent similar information through AVG, MEDI, and MODE versions, they are excluded from further EDA. Variables such as EXT_SOURCE_1 and OWN_CAR_AGE, although having high missing percentages, are retained for further consideration because their missingness may itself be meaningful or the variables may carry important predictive information.

```python
pd.crosstab(
    app_df["TARGET"],
    app_df["EXT_SOURCE_1"].isnull(),
    normalize="index"
) * 100
```

The missing percenatge of EXT_SOURCE_1 is 56.1 for Non-Defaulters and 59.5 for Defaulters. So we cannot conclude that missing values themselves indicate a higher default risk.  We will keep this column for now, but will delete property/building related columns with high missing values. Also, OWN_CAR_AGE also has high missing values, but we will leav eit as is now as many people might not have answered it as they do not own a car.

```python
cols_to_drop = [
    # Common Area
    'COMMONAREA_AVG', 'COMMONAREA_MEDI', 'COMMONAREA_MODE',

    # Non-living Apartments
    'NONLIVINGAPARTMENTS_AVG', 'NONLIVINGAPARTMENTS_MEDI', 'NONLIVINGAPARTMENTS_MODE',

    # Living Apartments
    'LIVINGAPARTMENTS_AVG', 'LIVINGAPARTMENTS_MEDI', 'LIVINGAPARTMENTS_MODE',

    # Floors Minimum
    'FLOORSMIN_AVG', 'FLOORSMIN_MEDI', 'FLOORSMIN_MODE',

    # Years Build
    'YEARS_BUILD_AVG', 'YEARS_BUILD_MEDI', 'YEARS_BUILD_MODE',

    # Land Area
    'LANDAREA_AVG', 'LANDAREA_MEDI', 'LANDAREA_MODE',

    # Basement Area
    'BASEMENTAREA_AVG', 'BASEMENTAREA_MEDI', 'BASEMENTAREA_MODE',

    # Non-living Area
    'NONLIVINGAREA_AVG', 'NONLIVINGAREA_MEDI', 'NONLIVINGAREA_MODE',

    # Elevators
    'ELEVATORS_AVG', 'ELEVATORS_MEDI', 'ELEVATORS_MODE',
]

app_df = app_df.drop(columns=cols_to_drop)

print(f"Shape after dropping columns: {app_df.shape}")
```

```python
contract_target = (
    pd.crosstab(app_df["NAME_CONTRACT_TYPE"],
                app_df["TARGET"],
                normalize="index")*100
)

contract_target
```

```python
contract_target.plot(
    kind="bar",
    stacked=True,
    figsize=(7,5)
)

plt.ylabel("Percentage")
plt.title("Default Rate by Contract Type")
plt.legend(["No Default","Default"])

plt.show()
```

Inference 1: Cash loan applicants exhibit a higher default rate (8.35%) compared to revolving loan applicants (5.48%). This suggests that contract type is a useful indicator of repayment risk and should be considered during loan approval and pricing decisions.

```python
categorical_cols = [
    "CODE_GENDER",
    "FLAG_OWN_CAR",
    "FLAG_OWN_REALTY",
    "NAME_INCOME_TYPE",
    "NAME_EDUCATION_TYPE",
    "NAME_FAMILY_STATUS",
    "NAME_HOUSING_TYPE",
    "OCCUPATION_TYPE",
    "ORGANIZATION_TYPE"
]

for col in categorical_cols:

    sns.countplot(data=app_df, x=col)
    plt.xticks(rotation=45, ha="right")
    plt.show()

    print("="*80)
    print(f"{col}")
    print("="*80)

    # Percentage table
    target_dist = pd.crosstab(
        app_df[col],
        app_df["TARGET"],
        normalize="index"
    ) * 100

    target_dist.columns = ["No Default (0)", "Default (1)"]

    display(target_dist.round(2))

    # Plot
    target_dist.plot(
        kind="bar",
        stacked=True,
        figsize=(8,4)
    )

    plt.title(f"Default Rate by {col}")
    plt.ylabel("Percentage")
    plt.xlabel(col)
    plt.xticks(rotation=45, ha="right")
    plt.legend(loc="upper right")
    plt.tight_layout()
    plt.show()

    default_rate = (
        pd.crosstab(app_df[col], app_df["TARGET"], normalize="index")[1]
        .sort_values(ascending=False) * 100
    )

    plt.figure(figsize=(8,4))

    sns.barplot(
        x=default_rate.index,
        y=default_rate.values
    )

    plt.title(f"Default Rate by {col}")
    plt.ylabel("Default Rate (%)")
    plt.xlabel(col)
    plt.xticks(rotation=45, ha="right")

    plt.tight_layout()
    plt.show()
```

Inference 2: 

Female applicants constitute a significantly larger proportion of the loan applicants, indicating that women apply for loans more frequently than men. However, despite being fewer in number, male applicants exhibit a higher default rate (10.14%) compared to female applicants (7.00%). This suggests that while women form the majority of the customer base, male applicants represent a relatively riskier borrower segment.

Inference 3: 

The majority of loan applicants belong to the Working category, followed by Commercial Associates. Among these major applicant groups, Working applicants exhibit the highest default rate (9.59%), indicating that they are relatively more prone to payment difficulties than other large income groups.

Although Maternity Leave (40%) and Unemployed (36.36%) show very high default rates, these categories contain very few applicants. Therefore, these percentages are based on a limited number of observations and should be interpreted cautiously rather than being considered strong evidence of higher risk.

Pensioners and State Servants exhibit comparatively lower default rates (around 5–6%), suggesting that applicants with relatively stable or predictable income sources tend to have better repayment behaviour.

Inference 4: 

Higher educational qualifications are associated with lower default rates, making education level an important indicator of repayment behaviour.

Inference 5:

The Laborers category deserves special attention because it combines the highest number of applicants with a high default rate (10.58%). Unlike Low-skill Laborers, whose default rate is even higher (17.15%) but with fewer observations, Laborers represent a much larger customer segment. Consequently, this group contributes significantly to the overall default population and should be considered an important focus area for credit risk assessment

Inference 6:

Applicants employed in low-skilled occupations, such as Low-skill Laborers, Drivers, Security Staff, Laborers, and Cooking Staff, exhibit noticeably higher default rates compared to applicants in professional or managerial occupations.

Inference 7:

Applicants living in Rented Apartments and With Parents exhibit the highest default rates (12.31% and 11.70%, respectively), whereas applicants living in their own House/Apartment have a comparatively lower default rate (7.80%). This suggests that stable housing or home ownership may be associated with better repayment behaviour.

Inference 8:

Although most applicants own real estate, the default rates for applicants who own and do not own real estate are almost identical (7.96% vs. 8.32%). This suggests that home ownership, by itself, is not a strong indicator of loan repayment behaviour in this dataset.

Inference 9:

Although applicants who do not own a car exhibit a slightly higher default rate (8.50%) than those who own a car (7.24%), the difference is relatively small. Therefore, car ownership alone does not appear to be a strong indicator of repayment behaviour.

Inference 10:

Married applicants account for the largest proportion of loan applications and exhibit a relatively lower default rate (7.56%) compared to Single/Not Married (9.81%) and Civil Marriage (9.94%) applicants. This suggests that married applicants tend to demonstrate slightly better repayment behaviour, possibly reflecting greater financial stability or dual-income households.

```python
# Lets check the numerical features and their distribution with respect to the target variable. We will use boxplots to visualize the distribution of numerical features for each target class.

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# List your numerical columns here
numerical_cols = [
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE",
    "DAYS_BIRTH",
    "DAYS_EMPLOYED",
    "CNT_CHILDREN",
    "CNT_FAM_MEMBERS"
]

for col in numerical_cols:

    print("=" * 100)
    print(f"Feature: {col}")
    print("=" * 100)

    # -------------------------
    # Summary Statistics
    # -------------------------
    print("\nOverall Summary:")
    print(app_df[col].describe())

    print("\nTarget-wise Summary:")
    print(app_df.groupby("TARGET")[col].describe())

    # -------------------------
    # Create plots
    # -------------------------
    fig, axes = plt.subplots(2, 2, figsize=(16, 10))

    # Histogram
    sns.histplot(app_df[col], bins=50, kde=True, ax=axes[0, 0])
    axes[0, 0].set_title(f"Distribution of {col}")

    # Boxplot
    sns.boxplot(x=app_df[col], ax=axes[0, 1])
    axes[0, 1].set_title(f"Boxplot of {col}")

    # Boxplot by TARGET
    sns.boxplot(data=app_df, x="TARGET", y=col, ax=axes[1, 0])
    axes[1, 0].set_title(f"{col} by TARGET")

    # KDE by TARGET
    sns.kdeplot(
        data=app_df[app_df["TARGET"] == 0],
        x=col,
        label="No Default",
        fill=True,
        alpha=0.3,
        ax=axes[1, 1]
    )

    sns.kdeplot(
        data=app_df[app_df["TARGET"] == 1],
        x=col,
        label="Default",
        fill=True,
        alpha=0.3,
        ax=axes[1, 1]
    )

    axes[1, 1].set_title(f"KDE of {col} by TARGET")
    axes[1, 1].legend()

    plt.tight_layout()
    plt.show()
```
